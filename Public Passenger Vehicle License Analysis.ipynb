<h2 style="color:#163A5F;">2.1 Cleaning Column Names</h2>

<p style="color:#555555;">
Before handling data quality issues, the column names are checked for extra
whitespace or inconsistent formatting. The <b>Taxi Medallion License
Management</b> column was found to contain a trailing space in its name,
which can cause errors when referencing the column. All column names are
therefore stripped of leading/trailing spaces.
</p>


```python
# Remove leading/trailing whitespace from column names
df.columns = df.columns.str.strip()

df.columns.tolist()
```

<h2 style="color:#163A5F;">2.2 Checking Missing Values</h2>

<p style="color:#555555;">
Missing values are identified in each column to understand the completeness
of the dataset and determine whether appropriate preprocessing is required.
</p>

```python
# Display columns with missing values

missing_values = df.isnull().sum()

missing_values[missing_values > 0]
```

```python
# Calculate missing values percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage[missing_percentage > 0].sort_values(ascending=False)
```

<h2 style="color:#163A5F;">2.3 Handling Missing Values</h2>

<p style="color:#555555;">
The missing values identified in Section 2.2 are now treated based on the
nature of each column. Before deciding on a strategy, the
<b>Vehicle Type</b> column is examined to understand why columns such as
<b>Taxi Affiliation</b> and <b>Taxi Medallion License Management</b> have a
very high percentage of missing values.
</p>


```python
# Check the distribution of Vehicle Type
df['Vehicle Type'].value_counts()
```

<p style="color:#555555;">
The <b>Taxi Affiliation</b> and <b>Taxi Medallion License Management</b>
columns only apply to vehicles of type <b>Taxi</b>. Other vehicle types
(such as Livery or TNP vehicles) do not have a taxi affiliation or medallion,
so their missing values represent <b>"Not Applicable"</b> rather than
genuinely unknown information.
</p>


```python
# Taxi-specific columns: missing values represent "Not Applicable"
df['Taxi Affiliation'] = df['Taxi Affiliation'].fillna('Not Applicable')
df['Taxi Medallion License Management'] = df['Taxi Medallion License Management'].fillna('Not Applicable')
```

```python
# Fill missing categorical values with "Unknown"
categorical_cols = ['Vehicle Make', 'Vehicle Model',
                     'Address', 'City', 'State']

df[categorical_cols] = df[categorical_cols].fillna('Unknown')

df['Vehicle Color'] = df['Vehicle Color'].fillna(df['Vehicle Color'].mode()[0])
```

```python
# Fill missing numeric values with the median
df['Vehicle Model Year'] = df['Vehicle Model Year'].fillna(df['Vehicle Model Year'].median())
df['ZIP Code'] = df['ZIP Code'].fillna(df['ZIP Code'].median())
```

```python
# Verify that no missing values remain
df.isnull().sum()
```

<h2 style="color:#163A5F;">2.4 Checking and Removing Duplicate Records</h2>

<p style="color:#555555;">
In addition to checking for fully duplicated rows, the <b>Record ID</b>
column (a unique identifier for each license record) is checked separately,
since duplicate IDs would indicate repeated records even if some other
column values differ.
</p>


```python
# Check for fully duplicated rows
print("Duplicate rows:", df.duplicated().sum())

# Check for duplicate Record IDs
print("Duplicate Record IDs:", df['Record ID'].duplicated().sum())
```

<h2 style="color:#163A5F;">2.5 Removing Unnecessary Columns</h2>


```python
# Check unique values in the State column

df['State'].unique()
```

```python
# Remove redundant and non-informative columns

df.drop(columns=['Record ID', 'State'], inplace=True)
```

```python
df.columns.tolist()
```

<p style="color:#555555;">
The <b>Record ID</b> column contains information already available in
<b>Public Vehicle Number</b> and <b>Vehicle Type</b>. Therefore, it is
removed as a redundant column.
</p>

<p style="color:#555555;">
The <b>State</b> column contains only one unique value. Since it does not
provide any variation for analysis, it is also removed.
</p>

<h2 style="color:#163A5F;">2.6 Correcting Data Types</h2>

<p style="color:#555555;">
Some columns are stored using data types that do not reflect their true
nature. <b>Vehicle Model Year</b> is stored as a float even though model
years are whole numbers, and <b>ZIP Code</b> is stored as a float even
though ZIP codes are identifiers rather than numeric quantities that should
be summed or averaged.
</p>


```python
# Vehicle Model Year: convert to integer (no missing values remain)
df['Vehicle Model Year'] = df['Vehicle Model Year'].astype(int)

# ZIP Code: convert to integer, then to string since it is an identifier, not a quantity
df['ZIP Code'] = df['ZIP Code'].astype(int).astype(str)
```

```python
# Verify the corrected data types
df.dtypes
```

<h2 style="color:#163A5F;">2.7 Standardizing Inconsistent Values</h2>

<p style="color:#555555;">
Text columns such as <b>Vehicle Make</b>, <b>Vehicle Color</b>, and
<b>City</b> may contain inconsistent capitalization or extra spacing. These columns are standardized by trimming whitespace and
converting text to uppercase for consistency.
</p>


```python
# Standardize text columns: strip whitespace and convert to uppercase
text_cols = ['Vehicle Type', 'Status', 'Vehicle Make', 'Vehicle Model',
             'Vehicle Color', 'Vehicle Fuel Source', 'Wheelchair Accessible',
             'Company Name', 'Address', 'City']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.upper()
```

```python
# Confirm the values are now consistent
df['City'].unique()
```

<h2 style="color:#163A5F;">2.8 Outlier Detection and Treatment</h2>


```python
# Boxplot of Vehicle Model Year before treatment
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['Vehicle Model Year'])
plt.title('Boxplot of Vehicle Model Year (Before Treatment)')
plt.show()
```

```python
df['Vehicle Model Year'].describe()
```

```python
# --Handling Outliers using IQR method---

CURRENT_YEAR = 2026  # no vehicle can have a model year beyond the current year

Q1 = df["Vehicle Model Year"].quantile(0.25)
Q3 = df["Vehicle Model Year"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = min(Q3 + 1.5 * IQR, CURRENT_YEAR)

print(f"Lower bound: {lower_bound}, Upper bound: {upper_bound}")
outliers = df[(df["Vehicle Model Year"] < lower_bound) | (df["Vehicle Model Year"] > upper_bound)]
print(f"Number of outliers detected: {len(outliers)}")

df = df[(df["Vehicle Model Year"] >= lower_bound) & (df["Vehicle Model Year"] <= upper_bound)]
print(f"Remaining rows: {len(df)}")
```

```python
# Boxplot of Vehicle Model Year after treatment
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['Vehicle Model Year'])
plt.title('Boxplot of Vehicle Model Year (After Treatment)')
plt.show()
```

<h2 style="background-color:#0B3954; padding:18px; border-radius:8px; color:white;">📌 Key Insights & Conclusion</h2>

<div style="background-color:#EAF3F8; padding:22px; border-radius:12px; border-left:6px solid #163A5F;">

### Key Insights

<h3 style="color:#0077B6; border-bottom:2px solid #0077B6; padding-bottom:5px;">1. Univariate Findings (Q1–Q6)</h3>

1. **Taxis and liveries dominate the licensed vehicles.** Together they make up 83.1% of all 16,140 licensed vehicles (Taxi: 43.4%, Livery: 39.8%); every other category combined accounts for under 17%.

2. **Most licenses are active, but violations are a major share.** Active vehicles form the largest group, followed by vehicles in violation and inactive status, while the remaining statuses are small.

3. **The fleet is mainly recent model years.** The most common model years are 2023 (1,617 vehicles), 2014 (1,364) and 2024 (1,247), with older years having much lower counts.

4. **The fleet runs mainly on gasoline and hybrid.** Gasoline accounts for 60.88% and hybrid for 35.30%, while every other fuel source is under 2%.

5. **License-holding companies cluster geographically.** ZIP code 60630 alone accounts for 3,086 licenses, far more than any other ZIP, with the top 10 ZIP codes concentrated on Chicago's North and Northwest Side.

6. **Ownership is concentrated in a handful of companies.** City of Chicago BACP holds the most licenses (669), followed by Superior Air-Ground Ambulance Service (322). Every other company holds fewer than 130 licenses.

<h3 style="color:#1F6F8B; border-bottom:2px solid #1F6F8B; padding-bottom:5px;">2. Bivariate Findings (Q7–Q13)</h3>

7. **Livery vehicles are notably newer than other categories.** Their average model year is 2022.7, more than 12 years newer than charter/sightseeing vehicles, with taxis sitting closer to the middle (about 2015 on average).

8. **Taxis carry a dramatically higher violation rate than liveries.** 50.2% of taxi licenses (3,516 of 6,999) are in *Violation* status, versus 9.5% of liveries (609 of 6,417), over 5x higher. This is the strongest compliance signal in the dataset.

9. **Fuel source diverges sharply by vehicle type.** Taxis run 81.3% hybrid (5,691 of 6,999), while liveries run 99.2% gasoline. The treemap confirms that gasoline liveries form the largest segment of the fleet, followed by hybrid and gasoline taxis.

10. **Violation risk within taxis is concentrated in a few affiliations.** "Expired Affiliation" (96.8%) and "Dis-Affiliated" (88.3%) taxis are almost entirely in violation, and Taxi Town Affiliation LLC also stands out at 75.2%. Established affiliations such as City Service Taxi Association (87.7% active) and Flash Cab Co (86.7% active) maintain strong compliance.

11. **Vehicle color strongly tracks vehicle type.** Liveries are almost entirely black (6,393 of about 6,406), while taxis are mostly white (4,614) but show more variety, including yellow (449) and blue (216), likely reflecting branding conventions.

12. **Vehicle numbering is unrelated to vehicle age.** The correlation between Public Vehicle Number and Model Year is very weak (r = 0.065), a useful sanity check rather than a substantive finding.

<h3 style="color:#3FA7A3; border-bottom:2px solid #3FA7A3; padding-bottom:5px;">3. Multivariate Findings (Q14–Q18)</h3>

13. **Wheelchair accessibility is concentrated in taxis.** Accessible vehicles appear mainly in the Taxi category, across gasoline, hybrid, flex fuel and CNG, while most vehicles of every type and fuel source are not accessible.

14. **Vehicle age depends on both type and status.** Active livery vehicles have the highest average model year (2023), while taxis appear across almost every licensing status.

15. **Vehicle make depends on fuel source and vehicle type.** Ford is common among gasoline and flex-fuel vehicles, Toyota among gasoline and hybrid taxis, Mercedes among diesel and bio-diesel vehicles, and Tesla among electric taxis. The animated chart shows that some makes are more common in newer model years and others in older ones.

16. **Vehicle type and licensing status both vary by ZIP code.** Liveries are concentrated in 60630 (about 2,100 licenses, mostly active), while taxis are spread more evenly, with the most in 60641 (about 1,570) and 60656 (about 1,170). Taxi violations are highest in 60630, where over 70% of taxis are in violation, and lowest in 60659, where active taxis are the majority. Violation problems are therefore concentrated in specific taxi-heavy areas.

### Conclusion

This analysis of Chicago's 16,140 licensed public passenger vehicles was carried out at three levels.

- **Univariate analysis** showed a fleet dominated by taxis and liveries, running mainly on gasoline and hybrid, made up of mostly recent model years, and concentrated in a few ZIP codes and companies.
- **Bivariate analysis** showed that the two dominant categories differ sharply in compliance, fuel composition, vehicle age and color. The most policy-relevant finding is the taxi fleet's high violation rate (50.2%), driven disproportionately by a small number of "dis-affiliated" or "expired affiliation" taxis, alongside its much stronger hybrid adoption (81.3%) compared to liveries (0.4%).
- **Multivariate analysis** showed that these differences are not uniform across the city. Vehicle make, age, accessibility and licensing status depend on vehicle type, fuel source and location together, and taxi violations are concentrated in specific areas such as ZIP code 60630.

### Recommendations & Next Steps

- **Target enforcement toward dis-affiliated and expired-affiliation taxis**, where violation rates exceed 88%, a small, identifiable segment driving much of the fleet-wide violation rate.
- **Prioritize taxi-heavy ZIP codes**, such as 60630, where the share of taxis in violation is highest.
- **Extend hybrid and electric incentives to liveries**, where adoption lags far behind taxis.
- **Track wheelchair-accessible vehicle share over time** as a standing equity metric, since accessible vehicles are concentrated in taxis.
- **Cross-reference with trip-level data** (also published by the City of Chicago) to connect vehicle characteristics (type, age, fuel, affiliation) to actual utilization.
